It is possible for an individual with these exact details to exist. However, since a duplicate entry appears immediately after the current one, the situation is suspicious; it should be verified whether these are truly two different people or if an error occurred, resulting in a double entry.

```python
def iqr_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return Q1, Q3, IQR, lower, upper

for col in ['chol', 'trestbps', 'thalach', 'oldpeak']:
    Q1, Q3, IQR, lo, hi = iqr_outliers(df[col])
    n_out = ((df[col] < lo) | (df[col] > hi)).sum()
    print(f"{col}: Q1={Q1}, Q3={Q3}, IQR={IQR:.2f}   ,     fences=({lo:.2f}, {hi:.2f}): Values ​​below {lo:.2f} and above {hi:.2f} are outliers.     ,      Number of outliers={n_out}")
```

```python

category_labels = {
    "cp": {
        0: "Typical Angina",
        1: "Atypical Angina",
        2: "Non-anginal",
        3: "Asymptomatic"
    },
    "fbs": {
        0: "FBS < 120",
        1: "FBS > 120"
    },
    "restecg": {
        0: "Normal",
        1: "ST-T Abnormality",
        2: "LV Hypertrophy"
    },
    "exang": {
        0: "No",
        1: "Yes"
    },
    "slope": {
        0: "Upsloping",
        1: "Flat",
        2: "Downsloping"
    },
    "ca": {
        0: "0 Vessels",
        1: "1 Vessel",
        2: "2 Vessels",
        3: "3 Vessels",
        4: "4 Vessels"
    },
    "thal": {
        0: "Unknown",
        1: "Fixed Defect",
        2: "Normal",
        3: "Reversible Defect"
    }
}


_default_palette = [
    "#4C72B0",  
    "#DD8452",  
    "#55A868",  
    "#C44E52",
    "#8172B2",  
    "#937860",  
    "#DA8BC3",  
    "#8C8C8C",  
]

feature_colors = {
    col: _default_palette
    for col in ["cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]
}


categorical_cols = [
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]


for col in categorical_cols:


    counts = df[col].value_counts().sort_index()   #Count of each feature

    labels = [
        category_labels[col].get(value, str(value))
        for value in counts.index
    ]

    colors = feature_colors[col][:len(counts)]

    
    # Figure Making
    fig, axes = plt.subplots(
        3,
        2,
        figsize=(18, 16)
    )


   
    # Row 1: Display bar chart (count per category)

    bars = axes[0, 0].bar(
        labels,
        counts.values,
        color=colors
    )

    for bar in bars:
        height = bar.get_height()
        axes[0, 0].text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{int(height)}',
            ha='center',
            va='bottom',
            fontsize=10,
            fontweight='bold'
        )

    axes[0, 0].set_title(
        f"{col} - Overall Distribution",
        fontsize=12,
        fontweight='bold'
    )
    axes[0, 0].set_xlabel(col)
    axes[0, 0].set_ylabel("Count")



    # Row 1 - Grand Total (Legend next to the chart)

    total_1 = counts.values.sum()

    # Descending sort by percentage (highest percentage first)
    sorted_1 = sorted(
        zip(labels, counts.values, colors),
        key=lambda item: item[1],
        reverse=True
    )
    labels_1_sorted = [item[0] for item in sorted_1]
    pie_data_1 = [item[1] for item in sorted_1]
    colors_1_sorted = [item[2] for item in sorted_1]

    pie_legend_labels_1 = [
        f"{label}: {value} ({value/total_1*100:.1f}%)"
        for label, value in zip(labels_1_sorted, pie_data_1)
    ]

    wedges_1, texts_1, autotexts_1 = axes[0, 1].pie(
        pie_data_1,
        autopct="%1.1f%%",
        startangle=90,
        colors=colors_1_sorted,
        textprops={"fontsize": 9, "fontweight": "bold", "color": "white"}
    )

    axes[0, 1].legend(
        wedges_1,
        pie_legend_labels_1,
        title=f"{col} Categories",
        loc="center left",
        bbox_to_anchor=(1.0, 0.5),
        fontsize=9,
        title_fontsize=10
    )

    axes[0, 1].set_title(
        f"{col} - Overall Percentage",
        fontsize=12,
        fontweight='bold'
    )



    # Row 2 - Male / Female Bar


    gender_counts = (
        df.groupby([col, "sex"])
          .size()
          .unstack(fill_value=0)
    )

    x = range(len(gender_counts))
    width = 0.35

    male_values = gender_counts.get(0, 0)
    female_values = gender_counts.get(1, 0)

    bars_male = axes[1, 0].bar(
        [i - width / 2 for i in x],
        male_values,
        width=width,
        label="Male",
        color="#6BAED6"
    )

    bars_female = axes[1, 0].bar(
        [i + width / 2 for i in x],
        female_values,
        width=width,
        label="Female",
        color="#F4A6C1"
    )

    # Sum of men and women for calculating the intra-group percentage
    male_total = sum(male_values)
    female_total = sum(female_values)

    for bar in bars_male:
        height = bar.get_height()
        percent = (height / male_total * 100) if male_total > 0 else 0
        axes[1, 0].text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{int(height)}\n({percent:.1f}%)',
            ha='center',
            va='bottom',
            fontsize=8,
            fontweight='bold',
            color='#2171B5'
        )

    for bar in bars_female:
        height = bar.get_height()
        percent = (height / female_total * 100) if female_total > 0 else 0
        axes[1, 0].text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{int(height)}\n({percent:.1f}%)',
            ha='center',
            va='bottom',
            fontsize=8,
            fontweight='bold',
            color='#C51B7D'
        )

    axes[1, 0].set_xticks(list(x))
    axes[1, 0].set_xticklabels(labels)

    axes[1, 0].set_title(
        f"{col} - Male vs Female",
        fontsize=12,
        fontweight='bold'
    )
    axes[1, 0].set_xlabel(col)
    axes[1, 0].set_ylabel("Count")
    axes[1, 0].legend()


    # Row 2 - Sex + Feature Pie 

    # Diverse color palette:
    male_palette = [
        "#1F77B4", 
        "#17BECF", 
        "#2CA02C", 
        "#9467BD",  
        "#003F5C",  
        "#00A6A6",  
        "#3F51B5",  
        "#4DBBD5",  
    ]

    female_palette = [
        "#E377C2",  
        "#FF7F0E",  
        "#D62728",  
        "#BC5090", 
        "#FFB6A3", 
        "#FFD92F",  
        "#E15759",  
        "#F28E8E",  
    ]

    pie_labels = []
    pie_values = []
    pie_colors_list = []

    for idx, feature_value in enumerate(gender_counts.index):
        label_name = labels[idx]
        male_val = gender_counts.loc[feature_value, 0] if 0 in gender_counts.columns else 0
        female_val = gender_counts.loc[feature_value, 1] if 1 in gender_counts.columns else 0

        # Male → From the cool palette
        pie_labels.append(f"Male - {label_name}")
        pie_values.append(male_val)
        pie_colors_list.append(male_palette[idx % len(male_palette)])

        # Female → From the warm palette
        pie_labels.append(f"Female - {label_name}")
        pie_values.append(female_val)
        pie_colors_list.append(female_palette[idx % len(female_palette)])


    if len(pie_values) <= 10:

        total_2 = sum(pie_values)

       # Descending sort by percentage (highest percentage first)
        sorted_2 = sorted(
            zip(pie_labels, pie_values, pie_colors_list),
            key=lambda item: item[1],
            reverse=True
        )
        pie_labels_sorted = [item[0] for item in sorted_2]
        pie_values_sorted = [item[1] for item in sorted_2]
        pie_colors_sorted = [item[2] for item in sorted_2]

        pie_legend_labels_2 = [
            f"{label}: {value} ({value/total_2*100:.1f}%)"
            for label, value in zip(pie_labels_sorted, pie_values_sorted)
        ]

        wedges_2, texts_2, autotexts_2 = axes[1, 1].pie(
            pie_values_sorted,
            autopct="%1.1f%%",
            startangle=90,
            colors=pie_colors_sorted,  #Diverse color palette:
            textprops={"fontsize": 8, "fontweight": "bold", "color": "white"},
            wedgeprops={"edgecolor": "white", "linewidth": 1.5}
        )

        axes[1, 1].legend(
            wedges_2,
            pie_legend_labels_2,
            title=f"{col} + Sex",
            loc="center left",
            bbox_to_anchor=(1.0, 0.5),
            fontsize=8,
            title_fontsize=9
        )

        axes[1, 1].set_title(
            f"{col} - Sex + Feature Distribution (%)",
            fontsize=12,
            fontweight='bold'
        )

    else:
        axes[1, 1].axis("off")
        axes[1, 1].text(
            0.5, 0.5,
            f"Too many categories ({len(pie_values)} slices)\nSee bar chart instead",
            ha="center", va="center",
            fontsize=12, style="italic", color="gray"
        )


    # Row 3 - Sex + Target Bar

    group_data = (
        df.groupby([col, "sex", "target"])
          .size()
          .unstack(["sex", "target"], fill_value=0)
    )

    group_order = [
        (0, 0),
        (1, 0),
        (0, 1),
        (1, 1)
    ]

    group_labels = [
        "Male - No Disease",
        "Female - No Disease",
        "Male - Disease",
        "Female - Disease"
    ]

    group_colors_list = [
        "#6BAED6",
        "#F4A6C1",
        "#2171B5",
        "#E31A1C"
    ]

    x = range(len(group_data))
    width = 0.18

    #We extract the values ​​for each group beforehand so that we can calculate the percentage relative to the total of each column.
    group_values_all = []
    for (sex, target) in group_order:
        if (sex, target) in group_data.columns:
            values = group_data[(sex, target)]
        else:
            values = [0] * len(group_data)
        group_values_all.append(list(values))

    # The sum of the values ​​of the four groups for each feature value (each x)
    totals_per_x = [
        sum(vals[j] for vals in group_values_all)
        for j in range(len(group_data))
    ]

    for i, ((sex, target), label, color) in enumerate(
        zip(group_order, group_labels, group_colors_list)
    ):
        values = group_values_all[i]

        bars = axes[2, 0].bar(
            [j + (i - 1.5) * width for j in x],
            values,
            width=width,
            label=label,
            color=color
        )

        for j, bar in enumerate(bars):
            height = bar.get_height()
            if height > 0:
                total = totals_per_x[j]
                percent = (height / total * 100) if total > 0 else 0
                axes[2, 0].text(
                    bar.get_x() + bar.get_width() / 2,
                    height,
                    f'{int(height)}\n({percent:.1f}%)',
                    ha='center',
                    va='bottom',
                    fontsize=7,
                    fontweight='bold'
                )

    axes[2, 0].set_xticks(list(x))
    axes[2, 0].set_xticklabels(labels)

    axes[2, 0].set_title(
        f"{col} - Sex + Target",
        fontsize=12,
        fontweight='bold'
    )
    axes[2, 0].set_xlabel(col)
    axes[2, 0].set_ylabel("Count")
    axes[2, 0].legend(fontsize=8)


    # The third row on the right is now empty.
    axes[2, 1].axis("off")



    # General Title
   

    fig.suptitle(
        f"Categorical Analysis of {col}",
        fontsize=16,
        fontweight="bold"
    )

    plt.tight_layout(
        rect=[0, 0, 1, 0.96]
    )

    plt.show()
```

```python

# Colors for gender and heart disease groups

group_colors = [
    "#6BAED6",   # Male - No Disease
    "#F4A6C1",   # Female - No Disease
    "#2171B5",   # Male - Disease
    "#E31A1C"    # Female - Disease
]

group_labels = [
    "Male - No Disease",
    "Female - No Disease",
    "Male - Disease",
    "Female - Disease"
]

# Create the figure and subplots

fig, axes = plt.subplots(
    3,
    2,
    figsize=(16, 15)
)

# First row: Heart disease distribution among all people

target_counts = (
    df["target"]
    .value_counts()
    .sort_index()
    .reindex([0, 1], fill_value=0)
)

target_labels = [
    "Has heart disease -",
    "Has heart disease +"
]

target_colors = [
    "#F5AFAF",
    "#F91D1D"
]

# Bar chart: Heart disease distribution among all people

ax = axes[0, 0]

bars = ax.bar(
    target_labels,
    target_counts.values,
    color=target_colors
)

ax.set_xlabel("Heart Disease")
ax.set_ylabel("Number of People")
ax.set_title("All People - Heart Disease")

# Display the number of people above each bar

for bar, value in zip(
    bars,
    target_counts.values
):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        str(value),
        ha="center",
        va="bottom"
    )

# Pie chart: Heart disease distribution among all people

ax = axes[0, 1]

ax.pie(
    target_counts.values,
    labels=target_labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=target_colors
)

ax.set_title(
    "All People - Heart Disease Distribution (%)"
)

# Second row: Heart disease distribution by gender

group_counts = [

    # Male without heart disease
    (
        (df["sex"] == 0) &
        (df["target"] == 0)
    ).sum(),

    # Female without heart disease
    (
        (df["sex"] == 1) &
        (df["target"] == 0)
    ).sum(),

    # Male with heart disease
    (
        (df["sex"] == 0) &
        (df["target"] == 1)
    ).sum(),

    # Female with heart disease
    (
        (df["sex"] == 1) &
        (df["target"] == 1)
    ).sum()
]

# Bar chart: Number of people by gender and heart disease status

ax = axes[2, 0]

# Define the positions of the bars

x = [0, 1, 2.5, 3.5]

bars = ax.bar(
    x,
    group_counts,
    width=0.8,
    color=group_colors
)

ax.set_title(
    "Heart Disease by Gender"
)

ax.set_xlabel("Gender + Heart Disease")
ax.set_ylabel("Number of People")

ax.set_xticks(x)
ax.set_xticklabels(
    group_labels,
    rotation=20
)

# Display the number of people above each bar

for bar, value in zip(bars, group_counts):

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        str(value),
        ha="center",
        va="bottom"
    )

# Pie chart: Percentage distribution by gender and heart disease status

ax = axes[2, 1]

ax.pie(
    group_counts,
    labels=group_labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=group_colors,
    textprops={"fontsize": 9}
)

ax.set_title(
    "Heart Disease by Gender (%)"
)

# Second row, left: Male and female distribution within each target group
# Display counts and percentages relative to each gender's total

# group_counts = [Male-0, Female-0, Male-1, Female-1]

male_t0, female_t0, male_t1, female_t1 = group_counts

total_t0 = male_t0 + female_t0
total_t1 = male_t1 + female_t1

male_values = [male_t0, male_t1]
female_values = [female_t0, female_t1]

# Calculate the total number of males and females for within-gender percentages

total_men = male_t0 + male_t1
total_women = female_t0 + female_t1

x = range(2)
width = 0.35

ax = axes[1, 0]

bars_male = ax.bar(
    [i - width / 2 for i in x],
    male_values,
    width=width,
    label="Male",
    color="#6BAED6"
)

bars_female = ax.bar(
    [i + width / 2 for i in x],
    female_values,
    width=width,
    label="Female",
    color="#F4A6C1"
)

# Display counts and percentages above the male bars

for i, bar in enumerate(bars_male):
    height = bar.get_height()
    percent = (height / total_men * 100) if total_men > 0 else 0
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}\n({percent:.1f}%)",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color="#2171B5"
    )

# Display counts and percentages above the female bars

for i, bar in enumerate(bars_female):
    height = bar.get_height()
    percent = (height / total_women * 100) if total_women > 0 else 0
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}\n({percent:.1f}%)",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color="#C51B7D"
    )

ax.set_xticks(list(x))
ax.set_xticklabels(target_labels)

ax.set_title(
    "Male vs Female in Each Target Group (% within sex)"
)
ax.set_xlabel("Heart Disease")
ax.set_ylabel("Number of People")
ax.legend()

# Hide the unused subplot in the second row, right

axes[1, 1].axis("off")

# Apply final layout adjustments

fig.suptitle(
    "Heart Disease Distribution",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout(
    rect=[0, 0, 1, 0.95]
)

plt.show()
```

```python
sns.set_theme(style="darkgrid")


# Create age groups

bins = [28, 38, 48, 58, 68, 78]

labels = [
    "29-38",
    "39-48",
    "49-58",
    "59-68",
    "69-78"
]

df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)


# Define features whose averages will be analyzed

num_cols = [
    "trestbps",
    "chol",
    "oldpeak",
    "thalach"
]


# Calculate the average for all people in each age group

age_group_mean = (
    df.groupby(
        "age_group",
        observed=True
    )[num_cols]
    .mean()
)

print("Average for all people:")
print(age_group_mean)


# Calculate the average separately for males and females in each age group

age_sex_mean = (
    df.groupby(
        ["age_group", "sex"],
        observed=True
    )[num_cols]
    .mean()
)

print("\nAverage by age group and sex:")
print(age_sex_mean)


# Create the plots

for col in num_cols:

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(16, 5)
    )


    # Left plot: Average for all people

    axes[0].bar(
        age_group_mean.index,
        age_group_mean[col]
    )

    axes[0].set_xlabel("Age Group")
    axes[0].set_ylabel(f"Average {col}")
    axes[0].set_title(
        f"Average {col} by Age Group"
    )


    # Right plot: Average for males and females

    male_mean = age_sex_mean[
        age_sex_mean.index.get_level_values("sex") == 0
    ][col]

    female_mean = age_sex_mean[
        age_sex_mean.index.get_level_values("sex") == 1
    ][col]


    # Convert the index so that age groups appear on the X-axis

    male_mean.index = male_mean.index.get_level_values(
        "age_group"
    )

    female_mean.index = female_mean.index.get_level_values(
        "age_group"
    )


    # Define the positions of the groups on the X-axis

    x = range(len(labels))

    width = 0.35


    # Male bars

    axes[1].bar(
        [i - width / 2 for i in x],
        male_mean.reindex(labels),
        width=width,
        label="Male"
    )


    # Female bars

    axes[1].bar(
        [i + width / 2 for i in x],
        female_mean.reindex(labels),
        width=width,
        label="Female"
    )


    axes[1].set_xticks(x)
    axes[1].set_xticklabels(labels)

    axes[1].set_xlabel("Age Group")
    axes[1].set_ylabel(f"Average {col}")

    axes[1].set_title(
        f"Average {col} by Age Group and Sex"
    )

    axes[1].legend()


    # Apply final layout adjustments

    plt.tight_layout()
    plt.show()
```
